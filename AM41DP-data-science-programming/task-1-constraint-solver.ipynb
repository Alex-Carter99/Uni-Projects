```python
#creating the initial array function
def Initial():
    a = np.zeros((5,5), dtype=object)
    for i in range(5):
        for j in range(5):
            a[i,j] = {1,2,3,4,5}
            
    r = np.zeros((5,4))
    
    c = np.zeros((4,5))
    
    return a, r, c
```

```python
#defining the total function
def Total(a):
    sum = 0
    for i in range(len(a)):
        for j in range(len(a[0])):
            sum = sum + len(a[i,j])
    return sum
```

```python
#defining the total function
def Uniqueness(a):
    for i in range(5):
        for j in range(5):
            if len(a[i,j]) == 1:
                single_value = 0
                for num in a[i,j]:
                    single_value = num
                
                # Remove this value from other cells in same row
                for col in range(5):
                    if col != j:
                        if single_value in a[i,col]:
                            a[i,col].remove(single_value)
                
                # Remove this value from other cells in same column
                for row in range(5):
                    if row != i:
                        if single_value in a[row,j]:
                            a[row,j].remove(single_value)
    
    return a
```

```python
#defining row inequalities function

def RowInequalities(a, r):
    for i in range(5):
        for j in range(4):  
            if r[i,j] == 1:  # Left > Right
               
                values_to_remove = set()
                for left_val in a[i,j]:
                    possible = False
                    for right_val in a[i,j+1]:
                        if left_val > right_val:  
                            possible = True
                            break
                    if not possible:
                        values_to_remove.add(left_val)
                
             
                if len(a[i,j] - values_to_remove) > 0:
                    a[i,j] -= values_to_remove
                
               
                values_to_remove = set()
                for right_val in a[i,j+1]:
                    possible = False
                    for left_val in a[i,j]:
                        if right_val < left_val:  
                            possible = True
                            break
                    if not possible:
                        values_to_remove.add(right_val)
                if len(a[i,j+1] - values_to_remove) > 0:
                    a[i,j+1] -= values_to_remove
                    
            elif r[i,j] == -1:  # Left < Right
                # Remove values from left cell that can never be less than right cell
                values_to_remove = set()
                for left_val in a[i,j]:
                    possible = False
                    for right_val in a[i,j+1]:
                        if left_val < right_val:  # Can this left value be less than some right value?
                            possible = True
                            break
                    if not possible:
                        values_to_remove.add(left_val)
                
                # Only remove if we're not emptying the cell
                if len(a[i,j] - values_to_remove) > 0:
                    a[i,j] -= values_to_remove
                
                # Remove values from right cell that can never be greater than left cell
                values_to_remove = set()
                for right_val in a[i,j+1]:
                    possible = False
                    for left_val in a[i,j]:
                        if right_val > left_val:  # Can this right value be greater than some left value?
                            possible = True
                            break
                    if not possible:
                        values_to_remove.add(right_val)
                
                # Only remove if we're not emptying the cell
                if len(a[i,j+1] - values_to_remove) > 0:
                    a[i,j+1] -= values_to_remove
    
    return a
```

```python
def ColumnInequalities(a, c):
    for i in range(4):
        for j in range(5):
            # First, check if any value appears only once in this column
            unique_values = set()
            for row in range(5):
                for val in a[row,j]:
                    count = sum(1 for r in range(5) if val in a[r,j])
                    if count == 1:
                        unique_values.add(val)
            
            if c[i,j] == 1:  # Top > Bottom
                values_to_remove = set()
                for top_val in a[i,j]:
                    possible = False
                    for bottom_val in a[i+1,j]:
                        if top_val > bottom_val:
                            possible = True
                            break
                    if not possible and top_val not in unique_values:
                        values_to_remove.add(top_val)
                
                if len(a[i,j] - values_to_remove) > 0:
                    a[i,j] -= values_to_remove
                
                values_to_remove = set()
                for bottom_val in a[i+1,j]:
                    possible = False
                    for top_val in a[i,j]:
                        if bottom_val < top_val:
                            possible = True
                            break
                    if not possible and bottom_val not in unique_values:
                        values_to_remove.add(bottom_val)
                
                if len(a[i+1,j] - values_to_remove) > 0:
                    a[i+1,j] -= values_to_remove
                    
            elif c[i,j] == -1:  # Top < Bottom
                values_to_remove = set()
                for top_val in a[i,j]:
                    possible = False
                    for bottom_val in a[i+1,j]:
                        if top_val < bottom_val:
                            possible = True
                            break
                    if not possible and top_val not in unique_values:
                        values_to_remove.add(top_val)
                
                if len(a[i,j] - values_to_remove) > 0:
                    a[i,j] -= values_to_remove
                
                values_to_remove = set()
                for bottom_val in a[i+1,j]:
                    possible = False
                    for top_val in a[i,j]:
                        if bottom_val > top_val:
                            possible = True
                            break
                    if not possible and bottom_val not in unique_values:
                        values_to_remove.add(bottom_val)
                
                if len(a[i+1,j] - values_to_remove) > 0:
                    a[i+1,j] -= values_to_remove
    
    return a
```

```python
def Pairs(a):
    # Check rows for pairs
    for i in range(len(a)):
        for j in range(len(a[0])):
            for k in range(j + 1, len(a[0])):
                if len(a[i,j]) == 2 and len(a[i,k]) == 2:
                    match = 1
                    for value in a[i,j]:
                        if value not in a[i,k]:
                            match = 0
                    for value in a[i,k]:
                        if value not in a[i,j]:
                            match = 0
                    if match == 1:
                        values_to_remove = []
                        for value in a[i,j]:
                            values_to_remove = values_to_remove + [value]
                        
                        for col in range(len(a[0])):
                            if col != j and col != k:
                                for value in values_to_remove:
                                    if value in a[i,col]:
                                        a[i,col].remove(value)
    # Check columns for pairs
    for j in range(len(a[0])):
        for i in range(len(a)):
            for k in range(i + 1, len(a)):
                if len(a[i,j]) == 2 and len(a[k,j]) == 2:
                    
                    match = 1
                    for value in a[i,j]:
                        if value not in a[k,j]:
                            match = 0
                    for value in a[k,j]:
                        if value not in a[i,j]:
                            match = 0
                    
                    if match == 1:
                        values_to_remove = []
                        for value in a[i,j]:
                            values_to_remove = values_to_remove + [value]
                        
                        for row in range(len(a)):
                            if row != i and row != k:
                                for value in values_to_remove:
                                    if value in a[row,j]:
                                        a[row,j].remove(value)
    
    return a
```

```python
def Triples(a):
    # Check rows for triples
    for i in range(len(a)):
        # Compare each combination of three cells in same row
        for j in range(len(a[0])):
            for k in range(j + 1, len(a[0])):
                for m in range(k + 1, len(a[0])):
                    # Check if all three cells have exactly 3 items
                    if len(a[i,j]) == 3 and len(a[i,k]) == 3 and len(a[i,m]) == 3:
                        # Check if cells contain same items
                        match = 1
                        for value in a[i,j]:
                            if value not in a[i,k] or value not in a[i,m]:
                                match = 0
                        for value in a[i,k]:
                            if value not in a[i,j] or value not in a[i,m]:
                                match = 0
                        
                        # If matching triple found, remove these values from other cells in row
                        if match == 1:
                            values_to_remove = []
                            for value in a[i,j]:
                                values_to_remove = values_to_remove + [value]
                            
                            for col in range(len(a[0])):
                                if col != j and col != k and col != m:
                                    for value in values_to_remove:
                                        if value in a[i,col]:
                                            a[i,col].remove(value)
    
    # Check columns for triples
    for j in range(len(a[0])):
        # Compare each combination of three cells in same column
        for i in range(len(a)):
            for k in range(i + 1, len(a)):
                for m in range(k + 1, len(a)):
                    # Check if all three cells have exactly 3 items
                    if len(a[i,j]) == 3 and len(a[k,j]) == 3 and len(a[m,j]) == 3:
                        # Check if cells contain same items
                        match = 1
                        for value in a[i,j]:
                            if value not in a[k,j] or value not in a[m,j]:
                                match = 0
                        for value in a[k,j]:
                            if value not in a[i,j] or value not in a[m,j]:
                                match = 0
                        
                        # If matching triple found, remove these values from other cells in column
                        if match == 1:
                            values_to_remove = []
                            for value in a[i,j]:
                                values_to_remove = values_to_remove + [value]
                            
                            for row in range(len(a)):
                                if row != i and row != k and row != m:
                                    for value in values_to_remove:
                                        if value in a[row,j]:
                                            a[row,j].remove(value)
    
    return a
```

```python
def Existence(a):
    # Check rows for unique values
    for i in range(len(a)):
        # Check each possible value (1-5)
        for value in range(1, 6):
            # Count occurrences of value in this row
            count = 0
            last_j = 0
            for j in range(len(a[0])):
                if value in a[i,j]:
                    count = count + 1
                    last_j = j
            
            # If value appears exactly once, make it the only value in its cell
            if count == 1:
                values_to_remove = []
                for item in a[i,last_j]:
                    if item != value:
                        values_to_remove = values_to_remove + [item]
                for item in values_to_remove:
                    a[i,last_j].remove(item)
    
    # Check columns for unique values
    for j in range(len(a[0])):
        # Check each possible value (1-5)
        for value in range(1, 6):
            # Count occurrences of value in this column
            count = 0
            last_i = 0
            for i in range(len(a)):
                if value in a[i,j]:
                    count = count + 1
                    last_i = i
            
            # If value appears exactly once, make it the only value in its cell
            if count == 1:
                values_to_remove = []
                for item in a[last_i,j]:
                    if item != value:
                        values_to_remove = values_to_remove + [item]
                for item in values_to_remove:
                    a[last_i,j].remove(item)
    
    return a
```

```python
def Main(a, r, c):
    # Initial setup
    T1 = Total(a)
    T2 = 126
    print("0", T1)
    
    iteration = 1
    while T1 > 25:
        # Apply uniqueness BEFORE other constraints
        a = Uniqueness(a)
        
        # Apply inequality constraints
        a = RowInequalities(a, r)
        a = ColumnInequalities(a, c)
        
        # Apply uniqueness after inequalities
        a = Uniqueness(a)
        
        # Apply other constraints
        a = Pairs(a)
        a = Triples(a)
        a = Existence(a)
        
        # Final uniqueness check after all constraints
        a = Uniqueness(a)
        
        # Calculate new total and continue as before
        T1 = Total(a)
        print(iteration, T1)
        
        # Error checking code
        if T1 > 125:
            print("Error: Too many possibilities")
            break
        
        if T1 < 25:
            print("Error: Too few possibilities")
            break
            
        if T1 == T2:
            print("Error: No progress made in this iteration")
            break
            
        T2 = T1
        
        if T1 == 25:
            print("Solution found:")
            for i in range(len(a)):
                for j in range(len(a[0])):
                    for value in a[i,j]:
                        print(value, end=" ")
                print()
            break
            
        iteration = iteration + 1
    
    return a
```
